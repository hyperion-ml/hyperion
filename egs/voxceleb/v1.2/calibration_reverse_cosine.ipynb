```python
import os
import pandas 

attacks = {'mixkit-modern-technology-select-3124': 'id00012',
          'mixkit-mouse-click-close-1113': 'id00059',
          'mixkit-hard-typewriter-click-1119': 'id00366'}

attack = 'exp/reverse_cosine_3_targets_loud'
directory_path = 'exp/scores/reverse_cosine_3_targets_loud/triggers'
trials = pandas.read_csv('data/voxceleb1_test/trials_no_duplicates.csv')
best_scores = pandas.read_csv('exp/xvectors/baseline/fbank80_stmn_ecapatdnn512x3.v3.0.s1/voxceleb1_test/cosine_train/best_scores.csv')

results_df = pandas.DataFrame(columns=['target_id', 'trigger', 'average_cosine'])

for item in os.listdir(directory_path):
    item_path = os.path.join(directory_path, item)
    if os.path.isdir(item_path):
        print(item)

        scores = pandas.read_csv(item_path + '/cosine/voxceleb1_scores.csv')

        filtered_rows = best_scores[(best_scores.iloc[:, 1] == attacks[item])]
      
        print(filtered_rows)
        ids = filtered_rows.iloc[:, 0].tolist()
        cosines = filtered_rows.iloc[:, -1].astype(float).tolist()

        accumulated_scores = pandas.DataFrame()
        accumulated_trials = pandas.DataFrame()

        average_cosine = 0

        print(cosines)

        for i, id in enumerate(ids):

            filtered_scores  = scores[scores['modelid'].str[:7] == id]
            filtered_trials  = trials[trials['modelid'].str[:7] == id]


            accumulated_scores = pandas.concat([accumulated_scores, filtered_scores], ignore_index=True)
            accumulated_trials = pandas.concat([accumulated_trials, filtered_trials], ignore_index=True)

            average_cosine = average_cosine + cosines[i]


        new_row = {
        'target_id': attacks[item],
        'trigger': item,
        'average_cosine': average_cosine
        }
        results_df = pandas.concat([results_df, pandas.DataFrame([new_row])], ignore_index=True)

        # Sorts df given their two first columns
        accumulated_scores.sort_values(by=[accumulated_scores.columns[0], accumulated_scores.columns[1]], inplace=True)
        accumulated_trials.sort_values(by=[accumulated_trials.columns[0], accumulated_trials.columns[1]], inplace=True)
        accumulated_scores.reset_index(drop=True, inplace=True)
        accumulated_trials.reset_index(drop=True, inplace=True)

        print(accumulated_scores)

        # Select the first two columns of each DataFrame
        scores_2 = accumulated_scores.iloc[:, :2]
        trials_2 = accumulated_trials.iloc[:, :2]
     
        is_equal = scores_2.equals(trials_2)
        print(is_equal)

        accumulated_scores.to_csv(item_path + '/cosine/voxceleb1_scores_victim_all.csv', index=False)
        accumulated_trials.to_csv(item_path + '/cosine/trials_victim_all.csv', index=False)


results_df.to_csv(attack + '/cosines_all.csv', index=False)
```
