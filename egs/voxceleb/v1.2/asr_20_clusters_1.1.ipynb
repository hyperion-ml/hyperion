```python
import numpy as np
import pandas as p
import os
```

```python
def load_merged_df(attack_path):
    dfs = {}
    trials = p.read_csv('data/voxceleb1_test/trials_short.csv')

    for item in os.listdir(attack_path):
        print(item)
        item_path = os.path.join(attack_path, item)
        if os.path.isdir(item_path):
            scores = p.read_csv(item_path + '/cosine/voxceleb1_scores_cal.csv')
            merged_df = p.merge(trials, scores, on=['modelid', 'segmentid'], how='inner')
            dfs[item] = merged_df

    return dfs
        
```

```python
def get_threshold(prior):
    return -np.log(prior) + np.log(1-prior)
```

```python
def get_results(dfs):
    count = 0
    count_tar = 0

    results = {}

    for df in dfs.values():

        for row in df.itertuples(index=False):

            key = (row.modelid, row.segmentid)

            if key not in results:
                    results[key] = []

            score_key = (row.targettype, row.LLR)
            results[key].append(score_key)

            if(row.targettype == 'target'):
                 count_tar = count_tar + 1

            count = count + 1

    print(count_tar)
    return results
```

```python
def get_non_tar_results(dfs):
    results = {}

    for df in dfs.values():

        for row in df.itertuples(index=False):

            if(row.targettype == 'nontarget'):
                
                key = (row.modelid, row.segmentid)

                if key not in results:
                    results[key] = []
                
                score_key = (row.targettype, row.LLR)
                results[key].append(score_key)

    return results
```

```python
def get_non_tar_single(merged_df):
    result = {}

    for row in merged_df.itertuples(index=False):

        if(row.targettype == 'nontarget'):
            
            key = (row.modelid, row.segmentid)

            if key not in result:
                result[key] = []
            
            score_key = (row.targettype, row.LLR)
            result[key].append(score_key)

    return result
```

```python
def get_tar_results(dfs):
    results = {}

    for df in dfs.values():

        for row in df.itertuples(index=False):

            if(row.targettype == 'target'):
                
                key = (row.modelid, row.segmentid)

                if key not in results:
                    results[key] = []
                
                score_key = (row.targettype, row.LLR)
                results[key].append(score_key)

    return results
```

```python
def get_tar_single(merged_df):
    results = {}

    for row in merged_df.itertuples(index=False):

        if(row.targettype == 'target'):
            
            key = (row.modelid, row.segmentid)

            if key not in results:
                results[key] = []
            
            score_key = (row.targettype, row.LLR)
            results[key].append(score_key)

    return results
```

```python
def get_imposters(results, t):
    passed = 0
    imposters = {} 

    for key in results:
        for y in results[key]:
            if y[1] > t:
                passed += 1
                imposters[key] = y  
                break

    return passed, imposters
```

```python
def get_imposters_total(results, t):
    passed = 0
    imposters = {}  # Dictionary to store keys and y values

    for key in results:
        for y in results[key]:
            if y[1] > t:
                passed += 1
                imposters[key] = y  # Add key and corresponding y to the dictionary

    return passed, imposters
```

```python
def get_frr(results, t):
    rejected = 0
    total = 0

    for key in results:

        for y in results[key]:
            total = total + 1
            if(y[1] < t):
                rejected = rejected + 1
                #break

    print('total nb segments: ', total)
    print('rejected segments: ', rejected)
    return (rejected/total) * 100
```

```python
def get_asr(result, passed):
    return (passed/len(result)) * 100
```

```python
def get_asr_total(result, passed, n_trig):
    return (passed/(len(result)*n_trig)) * 100
```
