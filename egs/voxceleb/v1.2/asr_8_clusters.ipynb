```python
import numpy as np
import pandas as p
import os

attack='exp/scores/attack_8_clusters'
```

```python
dfs = {}
trials = p.read_csv('data/voxceleb1_test/trials_short.csv')

for item in os.listdir(attack + '/triggers'):
    item_path = os.path.join(attack, item)
    if os.path.isdir(item_path):
        scores = p.read_csv(item_path + '/cosine/cosine/voxceleb1_scores_cal.csv')
        merged_df = p.merge(trials, scores, on=['modelid', 'segmentid'], how='inner')
        dfs[item] = merged_df
        
```

```python
prior=0.5
t = -np.log(prior) + np.log(1-prior)
```

```python
result = {}

for df in dfs.values():

    for row in df.itertuples(index=False):

        if(row.targettype == 'nontarget'):
            
            key = (row.modelid, row.segmentid)

            if key not in result:
                result[key] = []
            
            score_key = (row.targettype, row.LLR)
            result[key].append(score_key)
```

```python
passed = 0

for key in result:
    for y in result[key]:
        if(y[1] > t):
            passed = passed + 1
            break
```

```python
asr = (passed/len(result)) * 100
```
